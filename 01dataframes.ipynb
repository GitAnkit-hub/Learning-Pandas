Remember this trick
Think of the axis as the direction you move along:
- axis=0 → Move vertically (top to bottom).
- axis=1 → Move horizontally (left to right).

Most important: axis=0 means the row axis is reduced, while axis=1 means the column axis is reduced.

```python
students.mean(axis=1)
```

#### Selecting rows from a dataframe

* iloc - searches using index labels
* iloc - searches using index positions

```python
student_dict={
    'iq':[100,90,120,60,0,0],
    'marks':[80,70,114,100,0,0],
    'package':[10,7,15,9,0,0],
    'name':['nitish', 'Ankit', 'rupesh', 'rishabh', 'amit', 'ankita']
}

students = pd.DataFrame(student_dict)   # no need to mention columns explicity here.

students
```

```python
students.set_index('name', inplace=True)
```

```python
students
```

#Using iloc

```python
mvs.iloc[0]

# type(mvs.iloc[0]) # a series
```

```python
#multiple rows

mvs.iloc[[0,3,4]]  # using fancy indexing, also slicing

mvs.iloc[5:10]
mvs.iloc[4:14:2]
```

```python
students.loc[['Ankit', 'nitish']]
```
