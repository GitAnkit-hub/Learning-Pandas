#### Creating Dataframes

* single column is a series
* also, a single rows is a series

#Using 2Dlist

```python
student_list = [
    [100,80,10],
    [90,70,7],
    [120,114,15],
    [60, 100, 9]
]

pd.DataFrame(student_list, columns=['iq', 'marks', 'package'])
```

#Using Dict

```python
student_dict={
    'iq':[100,90,120,60],
    'marks':[80,70,114,100],
    'package':[10,7,15,9]
}

students = pd.DataFrame(student_dict)   # no need to mention columns explicity here.
```
