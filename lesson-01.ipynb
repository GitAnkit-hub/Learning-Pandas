## Creating Data 

    * DataFrame
    * Series

### DataFrame 

* A dataframe is a table. it contains an array of individual entries, each of which has a certain value. each entry corresponds to a row and a column.

```python
pd.DataFrame({'yes':[50,21], 'NO':[131, 2]})
```

```python
# pd.DataFrame is a constructor
# entries are not limited to integers.
```

#### Basically, A Pandas DataFrame is  a dictionary constructor where key is a column name and values are the list of entries.

```python
# row labels can be changed by using an "index" parameter in the constructor.

pd.DataFrame({'Name':["Ankit","Rohit", "suraj"], 'Age':[25, 26, 27]}, index = ['row0', 'row1', 'row2'])
```

### Series

* A series, by contrast, is a sequence of data values. if a dataframe is a table, a series a list. and in fact you can create one with nothing more than a list.

```python
pd.Series([1,2,34,5], name = "a series")

# you can give name of the series also
```

* A series is a single column of DataFrame.
* row label can be assigned using "index" parameter.
* A series does not have a column name

* A dataframe can be seen as multiple series glued together.
