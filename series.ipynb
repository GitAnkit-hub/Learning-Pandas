#### Series from lists

```python
#String

country = ['India', 'pakistan', 'Iran', 'Iraq', 'UK', 'USA']

pd.Series(country)
```

```python
#Intgers

runs = [23,43,3,65,7,43,32]

runs_series = pd.Series(runs)

```

```python
#custom Index

marks = [7,67,8,87,87]
subjects = ['history', 'hindi', 'geology', 'maths', 'mechanics']

pd.Series(marks, index=subjects)
```

```python
# Setting a name
marks = pd.Series(marks, index=subjects, name = 'ankit ke marks')

marks
```

#### Series from dictionary

```python
marks = {
    'history': 7,
    'hindi' :67,
    'gelology' : 8,
    'maths' : 87,
    'mechanics' : 87
}

marks_series = pd.Series(marks, name = 'mohit ke marks')
marks_series

# key automatically assign as index
```

```python
# mv[0] # this does not woek as the index is not integer, it is string
```

```python
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
```

```python
# s[0]  # same case
# # index is not integer
```

```python
s = pd.Series([10, 20, 30, 40],index=['a', 'b', 'c', 'd'])

s.loc['c']     # 30 (label)
s.iloc[2]      # 30 (position)

s.iloc[-1]     # 40 (last element)
# s.loc[-1]      # KeyError

s['c']         # 30 (direct label indexing)
```

* loc = Label, iloc = Integer Location (position).
