#### Indexing in Pandas

##### Indexing based selection
    * Selecting data based on its numerical position in the data.
    * returns a row of the table
    * df.iloc[row_position, column_position]
    * ":" means everything in python

```python
phase_review.iloc[92]
```

```python
print(phase_review.iloc[92, 3])  # row first, then column

print(phase_review['password'].iloc[92])  # first select password column then second its value
```

```python
# get complete column no. 3
phase_review.iloc[:, 3]
```

```python
phase_review.iloc[90]
```

```python
print(phase_review.iloc[-1:-6:-1]) # slicing concept also workingn

print(phase_review.iloc[-5:])
```

```python
phase_review.iloc[1:3, 0]
```

```python
phase_review.iloc[:, 0]
```

```python
phase_review.iloc[0:3, 0:3]
```

```python
# we can pass a list also

phase_review.iloc[[0,1,2], [0,1,2]]
```

```python
# A Range of rows can be selected
# A Range of cols can be selected
# Can be selected using either ":" or "list". 
```

##### Label based indexing
    * its data index value, not its position, which matters.
    * df.loc[row_label, column_label]
    * when column names are meaningful, its easier to do things with loc instead.

```python
phase_review.loc[2, 'password']
```

```python
# giving all rows for 2 selected columns
phase_review.loc[:, ['password', 'Name']]
```

##### iloc vs loc

```python
# in given range 0:10
    # iloc chooses 0 to 9
    # loc chooses 0 to 10
```

```python
df = pd.DataFrame({
    "Price": [100, 60, 80, 120]
}, index=["Apples", "Bananas", "Oranges", "Potatoes"])


df.loc["Apples":"Oranges"]


#This is convenient because you can specify the actual ending label.

df.iloc[:3]
```
