###`Q-2:` Write a Pandas program to add, subtract, multiple and divide two Pandas Series.

````python
# code here
L1 = [2,4,6,8,10]
L2 = [1,3,5,7,9]

s1 = pd.Series(L1)
s2 = pd.Series(L2)

print(s1.add(s2.values))

print(s1.subtract(s2.values))

print(s2.multiply(s2.values))

print(s2.divide(s2.values))

print(s1.values + s2.values)   # returns a list
````

````python
df.values
````

````python
def rupees(x):
    try:
        return float(x[1:])*94.25
    except:
        return float(x)*94.25

df = df.apply(rupees)
````

````python
df.dtype
````

````python
df = df.fillna(df.mean())

df.isnull().sum()
````
