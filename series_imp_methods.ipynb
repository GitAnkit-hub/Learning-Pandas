#drop_duplicates

```python
temp = pd.Series([1,2,2,1,3,4,4,5])
temp
```

```python
temp.drop_duplicates()

# phla occurence rakh liya aur baaki ke occurences delete kardiye

```

```python
# but we have the option to keep the last occurence also

temp.drop_duplicates(keep='last')
```

```python
temp.duplicated()

#true means-> ye value data me dusri baar aayi hai
#false means -> ye value data me phli baar aayi hai
```

```python
temp.duplicated().sum()  # 3 duplicated values hai

vk.duplicated().sum()  #137

mvs.duplicated().sum()  # 934

subs.duplicated().sum()  #185
```

```python
temp.drop_duplicates(keep='last').duplicated()  # returns a boolean series containing false
temp.drop_duplicates(keep='last').is_unique  # returns only True
```

#isnull

```python
temp = pd.Series([1,2,3,np.nan, 5,6,np.nan,8,np.nan,10])
temp
```

```python
temp.count()  # 7 -> counts only valid enrties

temp.size  #10
```

```python
temp.isnull()  #returns a boolean series

temp.isnull().sum()  # there are 3 nulls
```

#ab missing/null values ko 2 tariko se handle karskte hai

#dropna

```python
temp.dropna()  # removed all nan values

#returns a series
```

#fillna

```python
temp.fillna(0)  # replacing all the null values with 0

# returns a series


temp.fillna(temp.mean())
#temp.mean()->5
```

#copy
* A copy creates a separate Series with independent data.

```python
s1 = pd.Series([10,20,30])

s2 = s1.copy()

s2.iloc[0]=100

print(s1)
print(s2)
```

```python
#two types of copy

s2 = s1.copy(deep=True)   # Default  # a deep copy has independent data
s3 = s1.copy(deep=False)  #The original remains unchanged because of Copy-on-Write (CoW).
```

```python
s2[0]=100
s3[0]=100
```

```python
print(s2)
print(s3)
```
